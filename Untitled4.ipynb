```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing import image_dataset_from_directory

IMG_SIZE = (224, 224)
BATCH = 32
SEED = 123

train_ds = image_dataset_from_directory(
    "dataset",
    image_size=IMG_SIZE,
    batch_size=BATCH,
    validation_split=0.2,
    subset="training",
    seed=SEED
)

val_ds = image_dataset_from_directory(
    "dataset",
    image_size=IMG_SIZE,
    batch_size=BATCH,
    validation_split=0.2,
    subset="validation",
    seed=SEED
)

# Performance এর জন্য
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)

# এই লাইনটা আর দরকার নেই — বাদ দাও
# print("Class names:", train_ds.class_names)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Data Augmentation
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.05),
])

# Base Model
base = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base.trainable = False   # প্রথমে freeze

# Model
model = models.Sequential([
    data_augmentation,
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Data Augmentation বাদ — শুধু horizontal flip
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
])

# Base Model
base = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)
base.trainable = False

# Model
model = models.Sequential([
    data_augmentation,
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
class_weight = {0: 1.0, 1: 3.0}

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    class_weight=class_weight
)
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred_probs = []

for images, labels in val_ds:
    preds = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred_probs.extend(preds.flatten())

y_true = np.array(y_true)
y_pred_probs = np.array(y_pred_probs)

# Threshold 0.5 এ
y_pred = (y_pred_probs > 0.5).astype(int)

print("=== Threshold 0.5 ===")
print(confusion_matrix(y_true, y_pred))
print(classification_report(y_true, y_pred, target_names=["Normal", "Tuberculosis"]))
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred_probs = []

for images, labels in val_ds:
    preds = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred_probs.extend(preds.flatten())

y_true = np.array(y_true)
y_pred_probs = np.array(y_pred_probs)

# Threshold 0.5 এ
y_pred = (y_pred_probs > 0.5).astype(int)

print("=== Threshold 0.5 ===")
print(confusion_matrix(y_true, y_pred))
print(classification_report(y_true, y_pred, target_names=["Normal", "Tuberculosis"]))
```

```python
print("\n=== বিভিন্ন Threshold ===")
for threshold in [0.4, 0.5, 0.6, 0.7, 0.8]:
    y_pred_t = (y_pred_probs > threshold).astype(int)
    cm = confusion_matrix(y_true, y_pred_t)
    tn, fp, fn, tp = cm.ravel()
    tb_recall = tp / (tp + fn)
    normal_recall = tn / (tn + fp)
    tb_precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    print(f"Threshold {threshold}: TB Recall = {tb_recall:.3f}, Normal Recall = {normal_recall:.3f}, TB Precision = {tb_precision:.3f}")
```

```python
from tensorflow.keras.preprocessing import image
import numpy as np
import os

def predict_image(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)
    pred = model.predict(img_array, verbose=0)[0][0]
    label = "Tuberculosis" if pred > 0.5 else "Normal"
    print(f"{img_path}: {label} ({pred:.2%})")
    return pred

# Normal ছবি টেস্ট
print("=== Normal ছবি ===")
for i in range(3):
    normal_img = os.path.join("dataset/Normal", os.listdir("dataset/Normal")[i])
    predict_image(normal_img)

# TB ছবি টেস্ট
print("\n=== TB ছবি ===")
for i in range(3):
    tb_img = os.path.join("dataset/Tuberculosis", os.listdir("dataset/Tuberculosis")[i])
    predict_image(tb_img)
```
